```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca file csv
df = pd.read_csv('train.csv')

# Menampilkan 5 baris pertama
display(df.head())
```

```python
# Melihat apakah ada data yang kosong/missing values
print(df.isnull().sum())
```

```python
# Membuat Bar Chart keselamatan berdasarkan jenis kelamin
sns.countplot(data=df, x='Survived', hue='Sex')
plt.title('Perbandingan Keselamatan Penumpang (0 = Meninggal, 1 = Selamat)')
plt.show()
```
